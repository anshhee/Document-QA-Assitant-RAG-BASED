```python

import json
import numpy as np
import ollama
import chromadb
from pathlib import Path
import sys 
PROJECT_ROOT = Path.cwd().parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0 ,  str(PROJECT_ROOT))

cleaned_documents_file = Path("../data/processed/cleaned_documents.json")
chunked_documents_file = Path("../data/processed/chunked_documents.json")
embedding_file = Path("../data/processed/embeddings.npy")
chroma_path = Path("../data/chroma")

print("Setup complete")
```

```python
#loading documents

with open(cleaned_documents_file, "r", encoding="utf-8") as f:
    cleaned_documents = json.load(f)

print("Total documents:", len(cleaned_documents))
```

```python
#loading chunks 

with open(chunked_documents_file, "r", encoding="utf-8") as f:
    all_chunks = json.load(f)

print("Total chunks:", len(all_chunks))
```

```python

#loading embeddings 

embeddings_array = np.load(embedding_file)

print("Embeddings shape:", embeddings_array.shape)
```

```python
#checking if chunks and embeddings are loaded and matches
print("Chunks:", len(all_chunks))
print("Embeddings:", len(embeddings_array))

assert len(all_chunks) == len(embeddings_array)

print("Chunk and embedding counts match")
```

```python
#creating a chromadb client to store the embeddings

client = chromadb.PersistentClient(
    path=str(chroma_path)
)

print("ChromaDB client created successfully")
```

```python
#creating a collection to store into chromadb

collection = client.get_or_create_collection(
    name="documents"
)

print("Collection:", collection.name)
print("Current count:", collection.count())
```

```python
#storing the chunks in batched of size = 100

batch_size = 100

for start in range(0, len(all_chunks), batch_size):

    end = min(start + batch_size, len(all_chunks))

    batch_chunks = all_chunks[start:end]
    batch_embeddings = embeddings_array[start:end]

    collection.add(
        ids=[chunk["chunk_id"] for chunk in batch_chunks],
        embeddings=batch_embeddings.tolist(),
        documents=[chunk["text"] for chunk in batch_chunks],
        metadatas=[
            {
                "doc_id": chunk["doc_id"],
                "source": chunk["source"],
                "chunk_index": chunk["chunk_index"]
            }
            for chunk in batch_chunks
        ]
    )

    print(f"Inserted {end}/{len(all_chunks)} chunks")
```

```python
#checking the collection count

print("Chroma collection count:", collection.count())
```

```python
#Fetching the first chunk for verifying 

result = collection.get(
    ids=[all_chunks[0]["chunk_id"]],
    include=["embeddings", "documents", "metadatas"]
)

print("ID:", result["ids"][0])
print("Embedding dimensions:", len(result["embeddings"][0]))
print("Document:", result["documents"][0][:300])
print("Metadata:", result["metadatas"][0])
```

```python
def recall_at_k(retrieved_chunk_ids, relevant_doc_ids, k):
    """
    Calculate document-level Recall@K.

    retrieved_chunk_ids: ranked list of retrieved chunk IDs
    relevant_doc_ids: ground-truth parent document IDs
    k: number of top retrieved chunks to consider
    """

    # Take only the top K retrieved chunks
    top_k_chunks = retrieved_chunk_ids[:k]

    # Convert chunk IDs -> parent document IDs
    retrieved_doc_ids = [
        chunk_to_doc[chunk_id]
        for chunk_id in top_k_chunks
    ]

    # Deduplicate parent documents
    retrieved_docs = set(retrieved_doc_ids)
    relevant_docs = set(relevant_doc_ids)

    # Find relevant documents that were actually retrieved
    intersection = retrieved_docs & relevant_docs

    # Recall@K
    recall = len(intersection) / len(relevant_docs)

    return recall
```

```python
def precision_at_k(retrieved_chunk_ids, relevant_doc_ids, k):
    top_k_chunks = retrieved_chunk_ids[:k]

    retrieved_doc_ids = [
        chunk_to_doc[chunk_id]
        for chunk_id in top_k_chunks
    ]

    relevant_docs = set(relevant_doc_ids)

    relevant_retrieved = sum(
        1
        for doc_id in retrieved_doc_ids
        if doc_id in relevant_docs
    )

    precision = relevant_retrieved / k

    return precision
```

```python
def reciprocal_rank_fusion(
    dense_results,
    bm25_results,
    rrf_k=60
):
    scores = {}

    for rank, chunk_id in enumerate(dense_results, start=1):
        scores[chunk_id] = scores.get(chunk_id, 0) + (
            1 / (rrf_k + rank)
        )

    for rank, chunk_id in enumerate(bm25_results, start=1):
        scores[chunk_id] = scores.get(chunk_id, 0) + (
            1 / (rrf_k + rank)
        )

    ranked_results = sorted(
        scores.items(),
        key=lambda x: x[1],
        reverse=True
    )

    return [chunk_id for chunk_id, score in ranked_results]
```

```python
import re

def simple_rerank(query, candidate_ids):
    query_terms = set(
        re.findall(r"\b[a-zA-Z0-9]+\b", query.lower())
    )

    scored = []

    for chunk_id in candidate_ids:

        chunk = next(
            c for c in chunked_documents
            if c["chunk_id"] == chunk_id
        )

        text_terms = set(
            re.findall(r"\b[a-zA-Z0-9]+\b", chunk["text"].lower())
        )

        overlap = query_terms & text_terms

        score = len(overlap) / len(query_terms)

        scored.append((chunk_id, score))

    scored.sort(
        key=lambda x: x[1],
        reverse=True
    )

    return scored
```

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

model_name = "cross-encoder/ms-marco-MiniLM-L-6-v2"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)

model.eval()

print("Reranker loaded successfully")
```

```python
def rerank(query, candidate_chunks, top_k=5):
    pairs = [
        (query, chunk["text"])
        for chunk in candidate_chunks
    ]

    inputs = tokenizer(
        pairs,
        padding=True,
        truncation=True,
        max_length=512,
        return_tensors="pt"
    )

    with torch.no_grad():
        outputs = model(**inputs)

    scores = outputs.logits.squeeze(-1)

    ranked = sorted(
        zip(candidate_chunks, scores.tolist()),
        key=lambda x: x[1],
        reverse=True
    )

    return ranked[:top_k]
```

```python
def evaluate_reranker(k_values=(1, 3, 5, 10)):
    results = {
        k: []
        for k in k_values
    }

    for question_id, item in ground_truth.items():

        # Skip out-of-scope question
        if not item["source_doc_ids"]:
            continue

        query = item["question"]
        relevant_doc_ids = item["source_doc_ids"]

        # 1. Retrieve candidate pool
        candidate_ids = retrieve_candidate_pool(
            query,
            k=10
        )

        # 2. Convert chunk IDs → chunk objects
        candidate_chunks = [
            next(
                c for c in chunked_documents
                if c["chunk_id"] == chunk_id
            )
            for chunk_id in candidate_ids
        ]

        # 3. Rerank candidates
        reranked = rerank(
            query,
            candidate_chunks,
            top_k=10
        )

        # 4. Extract reranked chunk IDs
        reranked_ids = [
            chunk["chunk_id"]
            for chunk, score in reranked
        ]

        # 5. Calculate Recall@K
        for k in k_values:
            recall = recall_at_k(
                reranked_ids,
                relevant_doc_ids,
                k
            )

            results[k].append(recall)

    return results
```

```python
import json 
import ollama 


def evaluate_faithfullness(
    query,
    context,
    answer, 
    model = "qwen2.5:3b"
):

#prompt

    prompt = f"""
You are evaluating the faithfulness of a biomedical RAG answer.

Question:
{query}

Retrieved context:
{context}

Generated answer:
{answer}

Your task:
Determine whether the factual claims in the generated answer
are supported by the retrieved context.

For each factual claim:
1. Identify the claim.
2. Decide whether the retrieved context supports it.
3. Mark it as "supported", "partially_supported", or "unsupported".

Then calculate:

faithfulness_score =
number of fully supported claims / total factual claims

Rules:
- Use ONLY the retrieved context as evidence.
- Do not use your own medical knowledge.
- Do not reward a claim merely because it is medically plausible.
- If the context does not support a claim, mark it unsupported.
- If the context supports only part of a claim, mark it partially_supported.
- Do not penalize wording or style unless it changes the factual meaning.

Return ONLY valid JSON in this format:

{{
  "claims": [
    {{
      "claim": "...",
      "verdict": "supported"
    }}
  ],
  "faithfulness_score": 0.0,
  "reason": "..."
}}
"""

    response = ollama.generate(
        model=model,
        prompt=prompt
    )

    raw = response["response"]

    try:
        result = json.loads(raw)
    except json.JSONDecodeError:
        result = {
            "claims": [],
            "faithfulness_score": None,
            "reason": "Judge did not return valid JSON.",
            "raw_response": raw
        }

    return result
```

```python
def calculate_score_from_claims(result):

    claims = result.get("claims" , [])


    if not claims:
        return None

    supported_claims = sum(
        1
        for claim in claims
        if claim.get("verdict") == "supported"
    )

    total_claims = len(claims)

    return supported_claims/total_claims
    
```

```python
def evaluate_faithfulness_strict(
    query,
    context,
    answer,
    model="qwen2.5:3b"
):
    prompt = f"""
You are a strict evaluator of a biomedical RAG system.

Your task is to evaluate whether the GENERATED ANSWER is fully
supported by the RETRIEVED CONTEXT.

Question:
{query}

Retrieved context:
{context}

Generated answer:
{answer}

STEP 1 — Extract material factual claims.

Break the generated answer into individual factual claims.
Ignore:
- stylistic statements
- repetition
- opinions
- vague wording that adds no factual information

STEP 2 — Evaluate EVERY factual claim.

Use ONLY the retrieved context.

SUPPORTED:
The retrieved context explicitly supports the entire claim.

PARTIALLY_SUPPORTED:
The retrieved context supports only part of the claim,
but another part is unsupported or not established.

UNSUPPORTED:
The retrieved context does not provide evidence for the claim.

IMPORTANT:
- Do NOT use your own biomedical knowledge.
- Do NOT mark a claim supported because it is medically plausible.
- If a claim contains multiple factual assertions, ALL of them
  must be supported for the claim to be SUPPORTED.
- Extra information added by the generated answer must also be
  supported by the retrieved context.
- Be conservative.

Return ONLY valid JSON.

Use exactly this structure:

{{
  "claims": [
    {{
      "claim": "...",
      "verdict": "supported"
    }},
    {{
      "claim": "...",
      "verdict": "partially_supported"
    }},
    {{
      "claim": "...",
      "verdict": "unsupported"
    }}
  ]
}}
"""

    response = ollama.generate(
        model=model,
        prompt=prompt
    )

    raw = response["response"]

    try:
        result = json.loads(raw)
    except json.JSONDecodeError:
        result = {
            "claims": [],
            "raw_response": raw
        }

    return result
```

```python
#Generating answers for the questions from ground_truth.json


qwen3_generation_results = []

for question_id, item in ground_truth.items():

    # Q20 is out of scope and has no supporting documents
    if question_id == "20":
        continue

    print(f"Generating Q{question_id}...")

    result = generate_answer(
        query=item["question"],
        k=5
    )

    qwen3_generation_results.append({
        "question_id": question_id,
        **result
    })

print("\nCompleted:", len(qwen3_generation_results))
```
